# 3D Phase Portraits of Linear ODE Systems

**Elementary Differential Equations — Course Project**

This notebook visualizes solutions to the linear system **x′ = Ax** for five qualitatively distinct 3×3 matrices.
Trajectories are colored by instantaneous speed ‖x′(t)‖, encoding a 4th dimension as color.
Gold rays mark real eigenvectors; translucent cyan surfaces mark complex eigenplanes.

**Requirements:** Python 3.10+, VS Code with the [Jupyter extension](https://marketplace.visualstudio.com/items?itemName=ms-toolsai.jupyter), and run `pip install -r requirements.txt` before opening.

```python
import numpy as np
from solver import analyze, integrate_many
from renderer import plot_phase_portrait
from colormap_utils import compute_color_values_many
from demo_matrices import MATRICES, INITIAL_CONDITIONS, TITLES, sphere_ics
```

---
## 1 · Single figure

A stable spiral: eigenvalues −0.5 ± 2i and −1. One trajectory starting at (1, 0, 1) spirals
inward in the x₁-x₂ eigenplane while contracting along x₃.

```python
A = MATRICES['stable_spiral']
x0_list = [[1.0, 0.0, 1.0]]

_, eigen_info = analyze(A)
trajectories = integrate_many(A, x0_list, t_span=(0, 15), n_points=800)
color_values = compute_color_values_many(A, trajectories, quantity='speed')

fig = plot_phase_portrait(
    trajectories=trajectories,
    eigen_info=eigen_info,
    color_values=color_values,
    title=TITLES['stable_spiral'],
)
fig.show()
```

## 3 · 4D color encoding

The same stable spiral trajectory shown twice: left with speed coloring, right without.
Speed (‖x′(t)‖ = ‖Ax(t)‖) reveals that the trajectory slows as it approaches the origin —
information that position alone cannot show.

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

A = MATRICES['stable_spiral']
x0_list = INITIAL_CONDITIONS['stable_spiral']

_, eigen_info = analyze(A)
trajectories = integrate_many(A, x0_list, t_span=(0, 12))
color_values = compute_color_values_many(A, trajectories)

fig_color = plot_phase_portrait(
    trajectories, eigen_info, color_values,
    show_colorbar=True,
)
fig_plain = plot_phase_portrait(
    trajectories, eigen_info, color_values=None,
    line_color='steelblue',
    show_colorbar=False,
)

fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'scene'}, {'type': 'scene'}]],
    subplot_titles=('With speed coloring', 'Without color encoding'),
)

for trace in fig_color.data:
    fig.add_trace(trace, row=1, col=1)
for trace in fig_plain.data:
    fig.add_trace(trace, row=1, col=2)

fig.update_layout(
    title_text='4D encoding: speed ‖x′(t)‖ as color  —  Stable Spiral',
    height=550,
    scene=dict(xaxis_title='x₁', yaxis_title='x₂', zaxis_title='x₃', aspectmode='cube'),
    scene2=dict(xaxis_title='x₁', yaxis_title='x₂', zaxis_title='x₃', aspectmode='cube'),
)
fig.show()
```
